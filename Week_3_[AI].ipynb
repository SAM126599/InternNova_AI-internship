#1. What is Machine Learning?

Machine Learning (ML) is a branch of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every situation.

Instead of writing rules for every possible situation, we provide the computer with data. The ML algorithm studies the data, identifies patterns, and uses those patterns to make predictions on new data.

Example:
A student performance model can learn from study hours, attendance, previous marks, and assignment scores to predict whether a student is likely to Pass or Fail.

#📝 TASK 4 — Build Your First Machine Learning Model

Decision Tree Classifier

A Decision Tree is a supervised machine learning algorithm used for classification and regression.

```python
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load dataset
iris = load_iris()

# Create DataFrame
X = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

y = iris.target

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Create model
model = DecisionTreeClassifier(random_state=42)

# Train model
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Display predictions
print("Actual values:")
print(y_test)

print("\nPredicted values:")
print(y_pred)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

##Student Performance Prediction — Mini Project

This mini project focuses on predicting student performance using Machine Learning. A student performance dataset was created containing information about Study Hours, Attendance, Previous Marks, and Assignment Score. These variables were selected because they can provide useful information about a student's academic performance. The target variable in the dataset is Result, which contains two classes: Pass and Fail.

The dataset was loaded using the Pandas library and checked for missing values and data types. Any missing records were removed during the data-cleaning stage. The categorical target variable was converted into numerical values, where Fail was represented by 0 and Pass was represented by 1. The dataset was then divided into training and testing sets using an 80:20 ratio.

A Decision Tree Classifier from Scikit-learn was used to train the model. The model learned relationships between the selected student features and their final result. After training, predictions were generated using the testing dataset, and model accuracy was calculated using the accuracy_score() function.

The model was also tested with three new student inputs. A student with 6 study hours, 85% attendance, previous marks of 72, and an assignment score of 80 was predicted as PASS. Another student with lower academic indicators was predicted as FAIL.

The main limitation is that the dataset is small and may not represent real student behavior accurately. In the future, the project could be improved by using a larger real-world dataset, adding more features such as exam preparation, project performance, and participation, and comparing multiple classification algorithms.

#📚 Complete Mini-Project Code

```python
# ============================================
# STUDENT PERFORMANCE PREDICTION
# ============================================

import pandas as pd
import numpy as np
import io

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix, classification_report

# Define the raw CSV data as a string
csv_data = """
Study_Hours,Attendance,Previous_Marks,Assignment_Score,Result
8,95,88,92,Pass
6,85,72,80,Pass
7,90,80,85,Pass
3,65,50,55,Fail
2,60,45,50,Fail
9,96,92,95,Pass
5,78,65,70,Pass
4,70,55,60,Fail
10,98,95,96,Pass
1,55,40,45,Fail
6,88,75,82,Pass
3,62,48,52,Fail
8,92,85,90,Pass
5,75,60,68,Pass
2,58,42,48,Fail
7,86,78,84,Pass
4,68,52,58,Fail
9,94,90,93,Pass
6,82,70,76,Pass
3,64,46,54,Fail
8,89,82,87,Pass
2,57,44,46,Fail
5,80,68,72,Pass
4,72,58,62,Fail
10,97,94,98,Pass
7,91,81,86,Pass
1,52,35,40,Fail
6,84,73,79,Pass
3,66,49,56,Fail
9,93,89,91,Pass
"""

# 1. Load Dataset
df = pd.read_csv(io.StringIO(csv_data))

print("First 5 rows:")
print(df.head())


# 2. Dataset Information
print("\nDataset Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())


# 3. Clean Dataset
df = df.dropna()


# 4. Convert Target
df["Result"] = df["Result"].map({
    "Fail": 0,
    "Pass": 1
})


# 5. Select Features
X = df[
    [
        "Study_Hours",
        "Attendance",
        "Previous_Marks",
        "Assignment_Score"
    ]
]

# Target
y = df["Result"]


# 6. Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)


# 7. Create Model
model = DecisionTreeClassifier(
    random_state=42,
    max_depth=4
)


# 8. Train Model
model.fit(X_train, y_train)


# 9. Make Predictions
y_pred = model.predict(X_test)


print("\nActual Values:")
print(y_test.values)

print("\nPredicted Values:")
print(y_pred)


# 10. Calculate Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")


# 11. Confusion Matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))


# 12. Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))


# 13. Test New Students
new_students = pd.DataFrame({
    "Study_Hours": [6, 2, 8],
    "Attendance": [85, 60, 92],
    "Previous_Marks": [72, 45, 85],
    "Assignment_Score": [80, 50, 90]
})


predictions = model.predict(new_students)


# 14. Display Predictions
new_students["Prediction"] = np.where(
    predictions == 1,
    "PASS",
    "FAIL"
)


print("\nNew Student Predictions:")
print(new_students)
```
