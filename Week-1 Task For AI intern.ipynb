#📝 Task 3: Python for AI

Program 1: Student Pass/Fail Predictor (Classification AI Concept)

```python
# Pass/Fail Predictor
student_name = input("Enter student name: ")
marks = float(input("Enter exam score (0-100): "))

# Decision logic using operators
passing_score = 50.0

if marks >= passing_score:
    status = "PASSED"
    confidence = (marks / 100) * 100
else:
    status = "FAILED"
    confidence = ((100 - marks) / 100) * 100

print(f"\n[AI Result] Student: {student_name}")
print(f"Status: {status} (Model Confidence: {confidence:.1f}%)")
```

Program 2: Weather AI Temperature Converter

```python
# Temperature Converter for Weather AI
celsius_input = float(input("Enter temperature in Celsius: "))

# Conversion operators
fahrenheit = (celsius_input * 9/5) + 32
kelvin = celsius_input + 273.15

print("\n--- Weather AI Data Normalization ---")
print(f"Celsius:    {celsius_input:.2f} °C")
print(f"Fahrenheit: {fahrenheit:.2f} °F")
print(f"Kelvin:     {kelvin:.2f} K")
```

Program 3: AI Service Age Eligibility Checker

```python
# Age Eligibility Checker
user_age = int(input("Enter your age: "))

MIN_AGE_BASIC = 13
MIN_AGE_ADVANCED = 18

if user_age >= MIN_AGE_ADVANCED:
    access_level = "Full Access (Advanced AI Models & Generative Tools)"
elif user_age >= MIN_AGE_BASIC:
    access_level = "Restricted Access (Basic AI Tools with Safety Filters)"
else:
    access_level = "Access Denied (User below minimum age limit)"

print(f"\nAge: {user_age}")
print(f"Eligibility Result: {access_level}")
```

📝 Task 4: AI Case Study — Self-Driving Cars

```python
"""
Overview & How AI is Used:-
Self-driving cars (such as Tesla, Waymo, and Cruise) leverage Artificial Intelligence
to navigate roads safely without human intervention. The vehicle acts as an autonomous agent that perceives its environment,
makes real-time driving decisions, and executes physical controls (steering, braking, and acceleration).

1. Input (Perception Layer)
Autonomous vehicles continuously gather high-dimensional sensory data from multiple hardware sources:

LiDAR (Light Detection and Ranging): Fires laser beams to construct precise 3D spatial maps of the surroundings.

Cameras: High-resolution cameras capture visual data, including lane lines, traffic lights, and road signs.

Radar & Ultrasonic Sensors: Measure relative velocity and distance of surrounding objects under poor visibility conditions (rain, fog).

GPS & IMU: Provide location tracking and vehicle speed.

2. Processing (Cognition Layer)
The gathered data is processed by onboard edge-computing systems running Deep Learning pipelines:

Computer Vision (Convolutional Neural Networks - CNNs): Detects and classifies visual objects (pedestrians, cyclists, other vehicles, traffic signs).

Sensor Fusion: Combines raw inputs from LiDAR, Radar, and Cameras to create a unified 360-degree environmental model.

Path Planning Algorithms: Uses Reinforcement Learning and search algorithms (e.g., A
∗
  algorithm) to calculate optimal trajectories while anticipating movements of neighboring vehicles.

3. Output (Action Layer)
The output of the processing system translates into physical vehicle commands sent to Actuators:

Steering angle adjustments to stay centered in lanes or turn.

Brake pressure application when obstacles or traffic signals are detected.

Throttle adjustment to maintain speed limits or safely overtake cars.

4. Key Benefits
Safety: Over 90% of traffic accidents are caused by human error (distracted driving, fatigue, speeding). AI eliminates these human factors.

Accessibility: Provides independent mobility for the elderly and individuals with disabilities.

Traffic Optimization: AI-driven vehicles communicate to streamline traffic flow, reduce congestion, and lower energy consumption.
"""
```

📝 Task 5: Mini AI Project — Rule-Based Chatbot

```python
def rule_based_chatbot():
    print("==================================================")
    print("       🤖 Welcome to SimpleAI Chatbot 🤖         ")
    print("      (Type 'bye' or 'exit' to end conversation)   ")
    print("==================================================\n")

    bot_name = "Botty"

    while True:
        # Get user input and normalize text to lowercase
        user_input = input("You: ").strip().lower()

        # Exit condition
        if user_input in ["bye", "exit", "quit", "goodbye"]:
            print(f"{bot_name}: Goodbye! Have a great day ahead! 👋")
            break

        # Conversation Rules (if-elif-else logic)
        elif user_input in ["hello", "hi", "hey"]:
            print(f"{bot_name}: Hello! Welcome! How can I assist you today?")

        elif "how are you" in user_input:
            print(f"{bot_name}: I'm doing great, thank you for asking! How are you doing?")

        elif "your name" in user_input or "who are you" in user_input:
            print(f"{bot_name}: My name is {bot_name}, your friendly rule-based assistant.")

        elif "what is ai" in user_input or "define ai" in user_input:
            print(f"{bot_name}: AI stands for Artificial Intelligence—machines programmed to think and learn!")

        elif "python" in user_input:
            print(f"{bot_name}: Python is the primary language used for AI and Machine Learning!")

        elif "fine" in user_input or "good" in user_input:
            print(f"{bot_name}: That's wonderful to hear!")

        elif user_input == "":
            print(f"{bot_name}: It seems you didn't type anything. Feel free to ask me a question!")

        else:
            print(f"{bot_name}: I'm sorry, I don't understand that yet. Try asking me 'What is AI?' or say 'Hello'.")

# Execute the chatbot
if __name__ == "__main__":
    rule_based_chatbot()
```
